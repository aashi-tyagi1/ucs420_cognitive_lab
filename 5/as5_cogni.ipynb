```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

# a)
print("After adding 2:", arr + 2)

# b) 
print("After multiplying by 3:", arr * 3)

# c) 
print("After dividing by 2:", arr / 2)
```

```python

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("Most frequent values:", most_frequent)

for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("1st row, 2nd column : ", arr[0, 1])
print("2nd row, 3rd column : ", arr[1, 2])
```

```python
aashi = np.linspace(10, 100, 25)
print("Array:")
print(aashi)
print("Dimensions:", aashi.ndim)
print("Shape:", aashi.shape)
print("Total elements:", aashi.size)
print("Data type:", aashi.dtype)
print("Total bytes:", aashi.nbytes)
transposed = aashi.reshape(1, -1)
print("Transpose using reshape():")
print(transposed)
print("Transpose using T:")
print(aashi.T)
```

```python
ucs420_aashi = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])

print("Original array:")
print(ucs420_aashi)

print("Mean:", np.mean(ucs420_aashi))
print("Median:", np.median(ucs420_aashi))
print("Maximum:", np.max(ucs420_aashi))
print("Minimum:", np.min(ucs420_aashi))
print("Unique elements:", np.unique(ucs420_aashi))

reshaped_ucs420_aashi = ucs420_aashi.reshape(4, 3)

print("\nReshaped array:")
print(reshaped_ucs420_aashi)

resized_ucs420_aashi = np.resize(ucs420_aashi, (2, 3))

print("\nResized array:")
print(resized_ucs420_aashi)
```
