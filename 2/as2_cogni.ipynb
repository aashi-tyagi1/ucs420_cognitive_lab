# Tuples

```python
scores = L[0:9]
scores = tuple(scores)

highest = max(scores)
ind = scores.index(highest)

lowest = min(scores)
c = scores.count(lowest)

print(highest, ind,"\n" , sep = "    ")
print(lowest, c, sep = "    ")
```

# Random module

```python
import random
random.seed(1024170263)
R = []
for num in range(0,100):
    R.append(random.randint(100,900))
```

```python
c_odd = 0
c_eve = 0
E = []
O = []

for num in R:
    if(num%2 == 0):
        c_eve = c_eve +1
        E.append(num)
    else : 
        c_odd = c_odd +1
        O.append(num)
```

```python
print("Odd :- ", c_odd, O,sep = "\n")
print("Even :- ", c_eve, E,sep = "\n")
```

```python
def isPrime(n):
    if n <= 1:
        return False

    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1

    return True
```

```python
prime = 0
P = []


for num in R:
    if isPrime(num):
        P.append(num)
        prime += 1

P
```
